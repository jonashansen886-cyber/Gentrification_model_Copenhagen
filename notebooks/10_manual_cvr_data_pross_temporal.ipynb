```python
# change "id" to "CVREnhedsId" in all datasets for easier merging later
for dataset in [cvr_adress_cph, cvr_branche, cvr_navn, cvr_produktionsenhed, cvr_virksomhed]:
    for record in dataset:
        if "id" in record:
            record["CVREnhedsId"] = record.pop("id")
```

```python
# change names in datasets to not overlap

for record in cvr_navn:
    if "vaerdi" in record:
        record["navn"] = record.pop("vaerdi")

for record in cvr_branche:
    if "vaerdi" in record:
        record["branche_kode"] = record.pop("vaerdi")

for record in cvr_branche:
    if "vaerdiTekst" in record:
        record["branche_navn"] = record.pop("vaerdiTekst")
        

        

        
```

```python


# Free the unfiltered address list – only cvr_adress_cph is needed from here on
del cvr_adress
gc.collect()
print("Freed cvr_adress from memory")
```

```python
# join the datasets on "CVREnhedsId" to create a master dataset for Copenhagen
# Memory-efficient: build each lookup and free the source list immediately



def to_lookup(dataset):
    lookup = {}
    for r in dataset:
        eid = r.get("CVREnhedsId")
        if eid is not None:
            lookup[eid] = {k: v for k, v in r.items() if k != "CVREnhedsId"}
    return lookup

# Build lookups one at a time, freeing each source list after
branche_lkp = to_lookup(cvr_branche)
del cvr_branche; gc.collect()

navn_lkp = to_lookup(cvr_navn)
del cvr_navn; gc.collect()

prod_lkp = to_lookup(cvr_produktionsenhed)
del cvr_produktionsenhed; gc.collect()

virk_lkp = to_lookup(cvr_virksomhed)
del cvr_virksomhed; gc.collect()

print("Lookups built, source lists freed")

# Start from cvr_adress_cph and left-join the others
joined_cph = []
for rec in cvr_adress_cph:
    eid = rec.get("CVREnhedsId")
    merged = dict(rec)
    merged.update(branche_lkp.get(eid, {}))
    merged.update(navn_lkp.get(eid, {}))
    merged.update(prod_lkp.get(eid, {}))
    merged.update(virk_lkp.get(eid, {}))
    joined_cph.append(merged)

# Free lookups and the address list now that join is done
del branche_lkp, navn_lkp, prod_lkp, virk_lkp, cvr_adress_cph
gc.collect()

print(f"joined_cph: {len(joined_cph)} records, {len(joined_cph[0].keys())} columns")
print(f"Sample keys: {list(joined_cph[0].keys())[:15]}")
```

```python
# filter objects to only include certain feautures for easier handling later
features_to_keep = [
    "CVREnhedsId",
    "Adresse",
    "branche_kode",
    "branche_navn",
    "navn",
    "CVRNummer",
    "virksomhedStartdato",
    "virksomhedOphoersdato",
    "pNummer",
    "tilknyttetVirksomhedsCVRNummer",
    "produktionsenhedStartdato",
    "produktionsenhedOphoersdato",
]

```

```python
# filter joined_cph to only include features in features_to_keep
joined_cph = [
    {k: v for k, v in rec.items() if k in features_to_keep}
    for rec in joined_cph
]

# count how many unique keys appear across all records
all_keys = set()
for rec in joined_cph:
    all_keys.update(rec.keys())
print(f"Filtered to {len(all_keys)} unique features across {len(joined_cph)} records")
print(f"Features present: {sorted(all_keys)}")
```

```python
# group production units by their parent virksomhed CVRNummer
from collections import defaultdict

grouped = defaultdict(lambda: {"virksomhed": {}, "produktionsenheder": []})

for rec in joined_cph:
    cvr_nr = rec.get("CVRNummer") or rec.get("tilknyttetVirksomhedsCVRNummer")
    if cvr_nr is None:
        continue

    # store virksomhed-level fields once
    if not grouped[cvr_nr]["virksomhed"]:
        grouped[cvr_nr]["virksomhed"] = {
            "CVRNummer": cvr_nr,
            "navn": rec.get("navn"),
            "virksomhedStartdato": rec.get("virksomhedStartdato"),
            "virksomhedOphoersdato": rec.get("virksomhedOphoersdato"),
            "branche_kode": rec.get("branche_kode"),
            "branche_navn": rec.get("branche_navn"),
        }

    # append produktionsenhed info if present
    if rec.get("pNummer"):
        grouped[cvr_nr]["produktionsenheder"].append({
            "CVREnhedsId": rec.get("CVREnhedsId"),
            "Adresse": rec.get("Adresse"),
            "pNummer": rec.get("pNummer"),
            "tilknyttetVirksomhedsCVRNummer": rec.get("tilknyttetVirksomhedsCVRNummer"),
            "produktionsenhedStartdato": rec.get("produktionsenhedStartdato"),
            "produktionsenhedOphoersdato": rec.get("produktionsenhedOphoersdato"),
        })

grouped = dict(grouped)
print(f"Grouped into {len(grouped)} unique virksomheder")
# show an example with multiple produktionsenheder
example = next((v for v in grouped.values() if len(v['produktionsenheder']) > 1), None)
if example:
    print(f"\nExample virksomhed: {example['virksomhed']['navn']}")
    print(f"  CVRNummer: {example['virksomhed']['CVRNummer']}")
    print(f"  Produktionsenheder: {len(example['produktionsenheder'])}")
    for pe in example['produktionsenheder'][:3]:
        print(f"    pNummer: {pe['pNummer']}, Adresse: {pe['Adresse']}")
```

## Get adresses from DAWA API

```python
# input joined json dataset
with open("../data/processed/cvr/cvr_cph_joined_manual_temporal.json", "r", encoding="utf-8") as f:
    master_data = json.load(f)

print(f"Type: {type(master_data).__name__}, {len(master_data)} virksomheder")

# peek at the first entry
first_key = next(iter(master_data))
print(f"\nFirst CVRNummer: {first_key}")
print(json.dumps(master_data[first_key], indent=2, ensure_ascii=False))
```

```python
import aiohttp
import asyncio
import nest_asyncio
nest_asyncio.apply()
import time
import statistics
```

```python
# --- CONFIG ---
MAX_VIRKSOMHEDER = None   # None = process all
CONCURRENCY = 25              # 5 concurrent = ~10 req/s (API limit)
BATCH_SIZE = 500
MAX_RETRIES = 3
TIMEOUT_SEC = 30
SAVE_INTERVAL = 5000         # auto-save progress every N PEs

sem = asyncio.Semaphore(CONCURRENCY)

# --- MONITORING ---
api_times = []
status_codes = {}
retry_count = 0

async def fetch_coordinates(session, adresse_id):
    """Single-step DAWA lookup: /adresser/{id} already returns coordinates."""
    global retry_count
    for attempt in range(MAX_RETRIES):
        try:
            async with sem:
                t_start = time.perf_counter()

                async with session.get(
                    f"https://api.dataforsyningen.dk/adresser/{adresse_id}",
                    timeout=aiohttp.ClientTimeout(total=TIMEOUT_SEC),
                ) as resp:
                    status_codes[resp.status] = status_codes.get(resp.status, 0) + 1
                    if resp.status != 200:
                        api_times.append(time.perf_counter() - t_start)
                        return None, None
                    data = await resp.json()

                koordinater = data.get("adgangsadresse", {}).get("adgangspunkt", {}).get("koordinater", [])
                api_times.append(time.perf_counter() - t_start)
                if len(koordinater) == 2:
                    return koordinater[0], koordinater[1]
                return None, None

        except (aiohttp.ClientError, asyncio.TimeoutError):
            retry_count += 1
            if attempt < MAX_RETRIES - 1:
                await asyncio.sleep(1 * (attempt + 1))
                continue
            return None, None
    return None, None


def save_checkpoint():
    """Save current master_data progress to JSON."""
    path = "../data/processed/cvr/cvr_cph_addresses.json"
    with open(path, "w", encoding="utf-8") as f:
        json.dump(master_data, f, ensure_ascii=False)
    return path


async def process_all():
    pe_tasks = []
    count = 0
    for cvr_nr, entry in master_data.items():
        if MAX_VIRKSOMHEDER is not None and count >= MAX_VIRKSOMHEDER:
            break
        for pe in entry.get("produktionsenheder", []):
            adresse_id = pe.get("Adresse")
            if adresse_id:
                pe_tasks.append((pe, adresse_id))
        count += 1

    total_pe = len(pe_tasks)
    print(f"Queued {total_pe:,} PEs from {count:,} virksomheder (CONCURRENCY={CONCURRENCY})")
    print(f"Using single-step lookup (1 API call per PE instead of 2)")

    failed = 0
    connector = aiohttp.TCPConnector(limit=CONCURRENCY)
    async with aiohttp.ClientSession(connector=connector) as session:
        for batch_start in range(0, total_pe, BATCH_SIZE):
            batch = pe_tasks[batch_start : batch_start + BATCH_SIZE]
            tasks = [fetch_coordinates(session, aid) for _, aid in batch]
            results = await asyncio.gather(*tasks, return_exceptions=True)

            batch_failed = 0
            for (pe, _), result in zip(batch, results):
                if isinstance(result, Exception) or result == (None, None):
                    pe["longitude"] = None
                    pe["latitude"] = None
                    failed += 1
                    batch_failed += 1
                else:
                    pe["longitude"], pe["latitude"] = result

            done = min(batch_start + BATCH_SIZE, total_pe)
            elapsed_so_far = time.perf_counter() - t0
            avg_ms = statistics.mean(api_times) * 1000 if api_times else 0
            p50_ms = statistics.median(api_times) * 1000 if api_times else 0
            rps = done / elapsed_so_far if elapsed_so_far > 0 else 0
            eta_min = ((total_pe - done) / rps / 60) if rps > 0 else 0
            print(f"  [{done:,}/{total_pe:,}] ok:{len(batch)-batch_failed} fail:{batch_failed} | p50:{p50_ms:.0f}ms avg:{avg_ms:.0f}ms | {rps:.1f} PE/s | retries:{retry_count} | ETA:{eta_min:.1f}min")

            # auto-save checkpoint
            if done % SAVE_INTERVAL < BATCH_SIZE:
                save_checkpoint()
                print(f"    💾 checkpoint saved ({done:,} PEs)")

    return count, total_pe, failed


t0 = time.perf_counter()
count, total_pe, failed = asyncio.get_event_loop().run_until_complete(process_all())
elapsed = time.perf_counter() - t0

# final save
save_checkpoint()

print(f"\nProcessed {count:,} virksomheder, {total_pe:,} produktionsenheder in {elapsed:.1f}s ({elapsed/60:.1f}min)")
print(f"Coordinates found: {total_pe - failed:,}, failed: {failed:,}")

if api_times:
    avg_ms = statistics.mean(api_times) * 1000
    p50_ms = statistics.median(api_times) * 1000
    p95_ms = sorted(api_times)[int(len(api_times) * 0.95)] * 1000
    max_ms = max(api_times) * 1000
    rps = total_pe / elapsed
    print(f"\n--- API Rate Monitor (CONCURRENCY={CONCURRENCY}) ---")
    print(f"  Total lookups:   {len(api_times):,} (1 HTTP call each)")
    print(f"  Wall-clock:      {rps:.1f} PE/sec")
    print(f"  API latency p50: {p50_ms:.0f}ms")
    print(f"  API latency avg: {avg_ms:.0f}ms")
    print(f"  API latency p95: {p95_ms:.0f}ms")
    print(f"  API latency max: {max_ms:.0f}ms")
    print(f"  Retries used:    {retry_count} ({retry_count*100//(total_pe or 1)}% of lookups)")
    print(f"  HTTP status codes: {status_codes}")
    if retry_count > total_pe * 0.05:
        print(f"\n  ⚠ High retry rate — consider lowering CONCURRENCY to {max(3, CONCURRENCY-2)}")
    elif avg_ms < 300:
        print(f"\n  ✓ API latency healthy — could try CONCURRENCY={CONCURRENCY+5}")
```

```python
# flatten to one feature per produktionsenhed and save as GeoJSON
features = []
for cvr_nr, entry in master_data.items():
    virk = entry.get("virksomhed", {})
    for pe in entry.get("produktionsenheder", []):
        lon = pe.get("longitude")
        lat = pe.get("latitude")
        if lon is None or lat is None:
            continue
        features.append({
            "type": "Feature",
            "geometry": {"type": "Point", "coordinates": [lon, lat]},
            "properties": {
                **virk,
                "CVREnhedsId": pe.get("CVREnhedsId"),
                "Adresse": pe.get("Adresse"),
                "pNummer": pe.get("pNummer"),
                "tilknyttetVirksomhedsCVRNummer": pe.get("tilknyttetVirksomhedsCVRNummer"),
                "produktionsenhedStartdato": pe.get("produktionsenhedStartdato"),
                "produktionsenhedOphoersdato": pe.get("produktionsenhedOphoersdato"),
            },
        })

geojson = {"type": "FeatureCollection", "features": features}

output_path = "../data/processed/cvr/CVR_geo_temporal.geojson"
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(geojson, f, ensure_ascii=False, indent=2)
print(f"Saved {len(features)} produktionsenheder as GeoJSON to {output_path}")
```
