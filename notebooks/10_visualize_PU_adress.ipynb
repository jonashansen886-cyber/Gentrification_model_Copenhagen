## Map production units on map

```python
import geopandas as gpd
import leafmap
import os
import json

# Load the CVR GeoJSON
geojson_path = os.path.join("..", "data", "processed", "cvr", "CVR_geo.geojson")
gdf = gpd.read_file(geojson_path)
print(f"Loaded {len(gdf)} production units")
print(f"CRS: {gdf.crs}")
gdf.head()
```

```python
# Use lon/lat columns already present in the data for map centering
center_lat = gdf["lat"].mean()
center_lon = gdf["lon"].mean()

m = leafmap.Map(center=(center_lat, center_lon), zoom=7)

# Convert a 4326 copy only for GeoJSON display (original gdf stays UTM 32N)
geojson_data = json.loads(gdf.to_crs(epsg=4326).to_json())

# Style: blue circle markers
def style_function(feature):
    return {
        "fillColor": "#1f77b4",
        "color": "#1f77b4",
        "weight": 2,
        "fillOpacity": 0.7,
    }

m.add_geojson(
    geojson_data,
    layer_name="PU Units (CVR)",
    style_function=style_function,
)

# Save and display
output_path = os.path.join("..", "results", "figures", "pu_units_map.html")
m.to_html(output_path)
print(f"Map saved to {output_path}")
m
```
